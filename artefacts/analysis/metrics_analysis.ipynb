```python
# Recompute the core metric table directly from per-model JSON files.
from pathlib import Path
import json
import pandas as pd

rows = []
for path in sorted(Path('artefacts/metrics').glob('*_metrics.json')):
    rows.append(json.loads(path.read_text(encoding='utf-8')))
raw_metrics = pd.DataFrame(rows)
raw_metrics.sort_values(['macro_f1', 'micro_f1'], ascending=False)[['model', 'macro_f1', 'micro_f1', 'weighted_f1', 'precision_micro', 'recall_micro']]
```
